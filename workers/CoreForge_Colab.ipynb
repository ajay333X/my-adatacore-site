```python
import getpass, os
token = getpass.getpass('Paste the one-time CoreForge worker token (cfw_...): ').strip()
if not token.startswith('cfw_'):
    raise ValueError('That does not look like a CoreForge worker token.')
os.environ['COREFORGE_WORKER_TOKEN'] = token
os.environ['COREFORGE_POLL_SECONDS'] = '15'
print('Worker token loaded securely into this Colab session.')
```
